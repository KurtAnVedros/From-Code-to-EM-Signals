# To compare the similarity using euclidean distance, we need the signal data to be the same lenght. As such, we will remove access information at the end as we assume we have knowledge of the length of time the benign program will take.

```python
def cutToLength(dataset, length):
    new_Dataset = np.zeros((dataset.shape[0],length))
    
    for i in range(dataset.shape[0]):
        temp = dataset[i]
        new_Dataset[i] = temp[:length]
    
    new_Dataset = np.array(new_Dataset)
    return new_Dataset
```

```python
from mpl_toolkits.mplot3d import Axes3D
from matplotlib import cm
import matplotlib.pyplot as plt
from matplotlib.ticker import LinearLocator
```

```python
def Graph_3D(info, name, Program, Type):
    params = {'legend.fontsize': 30,
              'figure.figsize': (30, 15),
             'axes.labelsize': 30,
             'axes.titlesize':35,
             'xtick.labelsize':30,
             'ytick.labelsize':30}
    plt.rcParams.update(params)

    TrainingRange = np.arange(5,-1,-1)
    SNRRange = np.arange(4,-1,-1)

    fig = plt.figure()
    ax = fig.add_subplot(111, projection='3d')

    X = TrainingRange
    Y = SNRRange
    X, Y = np.meshgrid(X, Y)
    Z = info

    # Plot the surface.
    surf = ax.plot_surface(X, Y, Z, cmap=cm.coolwarm, vmin=0.5, vmax=1,
                           linewidth=0, antialiased=False)

    # Customize the axis limits.
    ax.set_xlim(0, 5)
    ax.set_ylim(0, 4)
    ax.set_zlim(0.5, 1)


    ax.set_title("", y=0.99,fontweight="bold")
    ax.set_xlabel("Samples", labelpad=25,fontweight="bold")
    ax.set_ylabel("SNR", labelpad=25,fontweight="bold")
    ax.set_zlabel(name, labelpad=20,fontweight="bold")

    a = ax.zaxis.label.get_rotation()
    if a<90:
       a += 90
    ax.zaxis.label.set_rotation(a)

    b = ax.yaxis.label.get_rotation()
    if b<25:
       b -= 30
    ax.yaxis.label.set_rotation(b)


    ax.view_init(elev=25., azim=225)

    ax.zaxis.set_major_locator(LinearLocator(10))
    # A StrMethodFormatter is used automatically
    ax.zaxis.set_major_formatter('{x:.02f}')

    ax.set_xticks([0, 1, 2, 3, 4, 5])
    ax.set_xticklabels(["10", "100", "500", "1000", "1500", "2000"])

    ax.set_yticks([0, 1, 2, 3, 4])
    ax.set_yticklabels(["10", "15", "20", "25", "30"])

    ax.set_zticks([0.5, 0.6, 0.7, 0.8, 0.9, 1.0])
    ax.set_zticklabels(["0.5", "0.6", "0.7", "0.8", "0.9", "1.0"])

    ax.tick_params(axis='z', which='major', pad=10)

    # Add a color bar which maps values to colors.
    fig.colorbar(surf, shrink=0.5, aspect=5, pad = 0)

    ax.set_box_aspect(aspect=None, zoom=0.93)

    folder = "./Results/3D/" + Program + "_" + Type + "/"

    # Check whether the specified path exists or not
    isExist = os.path.exists(folder)
    if not isExist:
      # Create a new directory because it does not exist 
      os.makedirs(folder)

    plt.savefig(folder + name + ".png", bbox_inches='tight') 
    plt.show()
    plt.close()
```

```python
def obtain3DGraphs(AUC, ACC, F1, Program, Type):
    # AUC
    Graph_3D(AUC, "AUC", Program, Type)
    
    # ACC
    Graph_3D(ACC, "ACC", Program, Type)
    
    # F1
    Graph_3D(F1, "F1", Program, Type)
```

```python
## Method for autoencoder, real noisy training
```

```python
def autoencoder_Real(program_N, program_A, program, SNR, train_amount):   
    save_path = "./Results/AE/" + program + "_SNR_" + str(SNR) + "/" + str(train_amount) + "/"

    # Dataset
    train_N = np.copy(program_N[0:train_amount])
    test_N = np.copy(program_N[2000:3000])
    test_A = np.copy(program_A[0:1000])
    # Create the X_train, X_test, and y_test
    X_train = train_N.astype('float32')
    X_test = np.concatenate((test_N, test_A)).astype('float32')

    zeros = np.repeat(0,test_N.shape[0])
    ones = np.repeat(1,test_A.shape[0])
    y_test = np.concatenate((zeros,ones))

    # Semi-supervised AutoEncoder
    from numpy.random import seed
    seed(4)
    input_dim = X_train.shape[1]

    Input_layer=Input(shape=(input_dim,))


    ## Create model
    Input_layer=Input(shape=(input_dim,))
    #encoder
    encoding_dim = 100
    from keras import initializers
    #init = initializers.VarianceScaling(scale=1.0, mode='fan_in', distribution='normal', seed=42)
    #encoder
    encoder=Dense(units=100,activation='tanh', use_bias=False, 
                  activity_regularizer=regularizers.l1(10e-5))(Input_layer)
    encoder=BatchNormalization()(encoder)
    encoder=Dense(units=int(50), activation='relu')(encoder)
    encoder=Dense(units=int(25), activation='relu')(encoder)

    #decoder
    decoder=Dense(units=int(50), activation='relu')(encoder)
    decoder=Dense(units=int(100), activation='relu')(decoder)
    decoder=Dense(units=input_dim,activation='relu')(decoder)

    #model
    autoencoder=Model(inputs=Input_layer,outputs=decoder)
    encodermodel=Model(inputs=Input_layer,outputs=encoder)

    # Load best weights (no training as its already done)
    autoencoder.load_weights(save_path + "best_weigths.hdf5")

    # Predict
    predictions_AE=autoencoder.predict(X_test, verbose=0)

    # Find the mean square error
    mse=np.mean(np.power(X_test - predictions_AE, 2), axis=1)

    fpr, tpr, tresholds = metrics.roc_curve(y_test, mse)
    scores_AE=[]
    for treshold in tresholds:
        y_hat_AE = (mse > treshold).astype(int)
        scores_AE.append([metrics.accuracy_score(y_pred=y_hat_AE, y_true=y_test),
                          metrics.f1_score(y_pred=y_hat_AE, y_true=y_test),
                         metrics.recall_score(y_pred=y_hat_AE, y_true=y_test),
                         metrics.precision_score(y_pred=y_hat_AE, y_true=y_test, zero_division=0.0),
                         metrics.fbeta_score(y_pred=y_hat_AE, y_true=y_test, beta=2),
                         metrics.cohen_kappa_score(y1=y_hat_AE, y2=y_test)])

    roc_auc = metrics.auc(fpr, tpr)

    scores_AE = np.array(scores_AE)
    final_tresh = tresholds[scores_AE[:, 0].argmax()]
    y_hat_AE = (mse > final_tresh).astype(int)
    best_score_AE = scores_AE[scores_AE[:, 0].argmax(),:]
    accuracy_score_AE = best_score_AE[0]
    f1_score_AE = best_score_AE[1]
    recall_score_AE = best_score_AE[2]
    precision_score_AE = best_score_AE[3]
    fbeta_score_AE = best_score_AE[4]
    cohen_kappa_score_AE = best_score_AE[5]

    del autoencoder
    keras.backend.clear_session()
    
    return roc_auc, accuracy_score_AE, f1_score_AE
```

```python
## Method for kNN
```

```python
def kNN_Real(program_N, program_A, program, SNR, train_amount):
    folder = "./Results/kNN/" + program + "_SNR_" + SNR + "/"
    
    Training_Amount = train_amount

    NormalTrain = program_N[0:Training_Amount]
    NormalTest = program_N[2000:3000]
    AnomalyTest = program_A[0:1000]

    normalTrainInst = "alpha SNR-" + SNR + " TS-" + str(Training_Amount)
    normalTestInst = "alpha SNR-" +SNR
    anomalyInst = "alpha_sub SNR-" +SNR

    name = normalTrainInst + " vs " + anomalyInst

    path = folder + name + "/"

    kFold = 1

    neighbor = np.array([10])

    f1Best, accBest, aucBest = KNN_Exp.runExperiment_once(NormalTrain, NormalTest, AnomalyTest, normalTrainInst, 
                                                        normalTestInst, anomalyInst, name, path, neighbors = neighbor, numFold = kFold,
                                                    show_Info=False)
    return aucBest, accBest, f1Best
```

# Next we need to place the information into arrays for 3D graphs

```python
'''
Re-runs all programs to obtain the found results. Does not train or retrains models.

program_N_Train: array of training sets
program_N_Test: array of normal testing sets, only user with autoencoder AE
program_A: array of anomalous test sets.
program: name of the program
SNRRange: range of snrs given as strings.
trainingRange: array of training ranges given as numbers
Type: the type of anomaly detection method.
    0: Autoencoder_Real
    1: Autoencoder_Synth
    2: kNN_Real
'''
def getInfo(program_N_Trains, program_N_Tests, program_As, program, SNRRange, trainingRange, Type):
    AUCs = []
    ACCs = []
    F1s = []
    
    # Cycle through the datasets
    for d in range(len(program_N_Train_Reals)):
        program_N_Train = program_N_Trains[d]
        program_N_Test = program_N_Tests[d]
        program_A = program_As[d]
        
        SNR = SNRRange[d]
        print("Working on SNR: ", SNR)
        
        AUC_SNR = []
        ACC_SNR = []
        F1_SNR = []

        # Cycle through the training Range
        for train_amount in trainingRange:
            # Rerun the program and get results
            if Type==0:
                AUC, ACC, F1= autoencoder_Real(program_N_Train, program_A, program, SNR, train_amount)

            elif Type==1:
                AUC, ACC, F1= autoencoder_Synth(program_N_Train, program_N_Test, program_A, program, SNR, train_amount)

            elif Type==2:
                AUC, ACC, F1= kNN_Real(program_N_Train, program_A, program, SNR, train_amount)
            #print("Train_amount: ", train_amount)
            # Store SNR results
            AUC_SNR.append(AUC)
            ACC_SNR.append(ACC)
            F1_SNR.append(F1)

        # Store all results
        AUCs.append(AUC_SNR)
        ACCs.append(ACC_SNR)
        F1s.append(F1_SNR)
        
    return np.array(AUCs), np.array(ACCs), np.array(F1s)    
```

```python
'''
Method runs through all 3 types of programs to collect and saves the results
'''
def allAD(program_N_Train_Reals, program_N_Train_Synths, program_As, program, SNRRange, trainingRange, folder):
    print("###### AE-Real #######")
    AUCs, ACCs, F1s = getInfo(program_N_Train_Reals, program_N_Train_Reals, program_As, program, SNRRange, trainingRange, Type=0)
    np.save(folder + "AE_Real-Noisy_AUCs.npy", AUCs)
    np.save(folder + "AE_Real-Noisy_ACCs.npy", ACCs)
    np.save(folder + "AE_Real-Noisy_F1s.npy", F1s)
    obtain3DGraphs(np.array(AUCs), np.array(ACCs), np.array(F1s), Program=program, Type="AE")
    
    print("###### AE-Synth #######")
    AUCs, ACCs, F1s = getInfo(program_N_Train_Synths, program_N_Train_Reals, program_As, program, SNRRange, trainingRange, Type=1)
    np.save(folder + "AE_Synth-Clean_AUCs.npy", AUCs)
    np.save(folder + "AE_Synth-Clean_ACCs.npy", ACCs)
    np.save(folder + "AE_Synth-Clean_F1s.npy", F1s)
    obtain3DGraphs(np.array(AUCs), np.array(ACCs), np.array(F1s), Program=program, Type="AE-Synth")
    
    print("###### kNN-Real #######")
    AUCs, ACCs, F1s = getInfo(program_N_Train_Reals, program_N_Train_Reals, program_As, program, SNRRange, trainingRange, Type=2)
    np.save(folder + "kNN_Real-Noisy_AUCs.npy", AUCs)
    np.save(folder + "kNN_Real-Noisy_ACCs.npy", ACCs)
    np.save(folder + "kNN_Real-Noisy_F1s.npy", F1s)
    obtain3DGraphs(np.array(AUCs), np.array(ACCs), np.array(F1s), Program=program, Type="kNN")
    
```
