```python
def NormalizeData(data, Min, Max):
    return (data - Min) / (Max - Min)
```

```python
def rollingEuclidean_Peaks(FullSignal, SectionSignals, lowestMin, pad, amountPeaks):
    Size = SectionSignals[0].shape[0]
        
    rollingEuclideanList = []
    
    tempSect = MapTool.getPeaks(SectionSignals[0], pad)
    if MapTool.getPeaksLoc(SectionSignals[0], pad, tempSect.shape[0]-1) > Size - pad:
        peakLengthSec = MapTool.getPeaks(SectionSignals[0], pad).shape[0] -1 # 0 for the start 
    else:
        peakLengthSec = MapTool.getPeaks(SectionSignals[0], pad).shape[0] # 1 for the start and end points
    #print(peakLengthSec)
    
    tempSignal = FullSignal[:FullSignal.shape[0]-1-Size]
    peaks = MapTool.getPeaks(tempSignal , pad)
    
    for i in range(peaks.shape[0]):
        j = MapTool.getPeaksLoc(FullSignal, pad, i)
        k = MapTool.getPeaksLoc(FullSignal, pad, i+amountPeaks)
        
        #print("i", i)
        #print("i + amountPeaks", i+amountPeaks)
        
        part = FullSignal[j:k]
        
        part_Peaks = MapTool.getPeaks(part, pad)
        
        if part_Peaks.shape[0] + 1 == amountPeaks:
            part_Peaks = np.append(part_Peaks, part[part.shape[0]-1])
        if part_Peaks.shape[0] - 1 == amountPeaks:
            part_Peaks = np.delete(part_Peaks, part_Peaks.shape[0]-1)
        
        topCorr = -90000000000
        
        for s in range(SectionSignals.shape[0]):           
            sect = SectionSignals[s]
            SectSignal_Peaks = MapTool.getPeaks(sect, pad)
            if SectSignal_Peaks.shape[0] + 1 == amountPeaks:
                SectSignal_Peaks = np.append(SectSignal_Peaks, sect[sect.shape[0]-1])
            if SectSignal_Peaks.shape[0] - 1 == amountPeaks:
                print(SectSignal_Peaks)
                plotSignal("sect_Peaks", sect)
                print(s)
            
            #print(part_Peaks)
            #plotSignal("part_Peaks", part)
            #print(SectSignal_Peaks)
            #plotSignal("SectSignal_Peaks", sect)
            #print("part_Peaks" , part_Peaks.shape)
            #print("SectSignal_Peaks", SectSignal_Peaks.shape)
               
            distSect = distance.euclidean(part_Peaks, SectSignal_Peaks)
            distSect = distSect * -1
            
            if distSect > topCorr:
                topCorr = distSect
        
        rollingEuclideanList.append(topCorr)
        
    rollingEuclideanList = np.array(rollingEuclideanList, dtype=object)
    
    Min = lowestMin
    Max = 0
    
    rollingEuclideanBestCorrelation = NormalizeData(rollingEuclideanList, Min, Max)
    
    return rollingEuclideanBestCorrelation  
```

```python
def plotCorrelation(correlation, titleName):
    ### Plot the correlation
    plt.plot(correlation)
    plt.title(titleName)
    plt.xlabel("Index of Signal")
    plt.ylabel("Correlation")
    plt.show()
```

```python
def plotMostLikelyArea(Signal, Seq, euc_corr, signalName, seqName, titleName, pad):
    ### Plot the most likely area in the signal sample

    peakNum = np.argmax(euc_corr)
    loc = MapTool.getPeaksLoc(Signal, pad, peakNum)
    corr = max(euc_corr)
    diff = Seq.shape[0]
    SeqInSample = Signal[loc:loc+diff]

    time_Signal = np.arange(Signal.shape[0])
    time_Seq = np.arange(loc,loc+diff)

    plt.plot(time_Signal, Signal, label = signalName)
    plt.plot(time_Seq, SeqInSample, label = seqName)
    plt.title(titleName)
    plt.xlabel("Sample Index")
    plt.ylabel("Amplitude")
    plt.legend()
    plt.show()
```

```python
def getSeqSignals(Signals, peaksStart, peaksEnd, numPeaksWhenGiven):
    seqList = []

    for signal in Signals:    
        Seq = signal[MapTool.getPeaksLoc(signal, pad, peaksStart):MapTool.getPeaksLoc(signal, pad, peaksEnd)]
        seqList.append(Seq)
    
    seqList = np.array(seqList, dtype="object")
    return seqList
```

```python
def getEvaluations(corrs, ground_truths, numberThresholds=1000):
    results = []

    for threshold in np.arange(0, 1 + 1 / numberThresholds, 1 / numberThresholds):
        predictions = []
        for corr in corrs:
            for index in range(corr.shape[0]):
                if corr[index] > threshold:
                    predictions.append(1)
                else:
                    predictions.append(0)
        
        predictions = np.array(predictions)
        
        ground_truth = ground_truths.flatten()

        metrics = ConfidenceEstimation.calculateBasicMetrics(ground_truth, predictions)

        results.append(metrics)

    return results
```
